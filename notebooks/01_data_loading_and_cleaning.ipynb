```python
# Create a copy of the original dataset
clean_df = df.copy()

print(clean_df.shape)
```

```python
clean_df = clean_df.dropna(subset=["Customer ID"])

print(clean_df.shape)
```

```python
clean_df = clean_df.drop_duplicates()

print(clean_df.shape)
```

```python
clean_df = clean_df[
    ~clean_df["Invoice"].astype(str).str.startswith("C")
]

print(clean_df.shape)
```

```python
clean_df = clean_df[
    clean_df["Quantity"] > 0
]

print(clean_df.shape)
```

```python
clean_df = clean_df[
    clean_df["Price"] > 0
]

print(clean_df.shape)
```

```python
clean_df.isnull().sum()
```

```python
clean_df.to_csv(
    "../data/processed/cleaned_online_retail.csv",
    index=False
)
```

## Data Cleaning Summary

The dataset was cleaned using the following steps:

- Removed records with missing Customer IDs.
- Removed duplicate transactions.
- Removed cancelled invoices (Invoice starting with 'C').
- Removed transactions with non-positive quantities.
- Removed transactions with non-positive prices.

After cleaning, the dataset contains **779,425 valid customer transactions** and is ready for feature engineering.

```python
clean_df["InvoiceMonth"] = clean_df["InvoiceDate"].dt.to_period("M")
```

```python
clean_df["CohortMonth"] = (
    clean_df.groupby("Customer ID")["InvoiceDate"]
            .transform("min")
            .dt.to_period("M")
)
```

```python
invoice_year = clean_df["InvoiceDate"].dt.year
invoice_month = clean_df["InvoiceDate"].dt.month

cohort_year = clean_df["CohortMonth"].dt.year
cohort_month = clean_df["CohortMonth"].dt.month

clean_df["CohortIndex"] = (
    (invoice_year - cohort_year) * 12
    + (invoice_month - cohort_month)
    + 1
)
```

```python
clean_df["Year"] = clean_df["InvoiceDate"].dt.year
clean_df["Month"] = clean_df["InvoiceDate"].dt.month
clean_df["Day"] = clean_df["InvoiceDate"].dt.day
clean_df["DayOfWeek"] = clean_df["InvoiceDate"].dt.day_name()
clean_df["Hour"] = clean_df["InvoiceDate"].dt.hour
```

```python
clean_df["CustomerRevenue"] = (
    clean_df.groupby("Customer ID")["Revenue"]
            .transform("sum")
)
```

```python
clean_df["CustomerOrders"] = (
    clean_df.groupby("Customer ID")["Invoice"]
            .transform("nunique")
)
```

```python
clean_df["AverageOrderValue"] = (
    clean_df["CustomerRevenue"] /
    clean_df["CustomerOrders"]
)
```

```python
clean_df.to_csv(
    "../data/processed/feature_engineered_retail.csv",
    index=False
)
```
